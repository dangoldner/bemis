# Summary Notes of Chapters I've Read

**Ch 1 covered**: EMH, systemic vs idiosyncratic risk, CAPM (β as market sensitivity, α as excess return), Fama-French (adding size and value factors), and MVO setup. Key theme: models are useful for interpretation and comparison, but their stationarity assumptions are violated in practice.

**Ch 2 covered**: CDFs, PDFs, moments (mean, variance, skew, kurtosis), percentiles. Normal, log-normal, and Student *t* distributions — with emphasis on how real equity returns exhibit fat tails and asymmetry that the normal badly misses (837,128× underestimate of 6σ events!). Student *t* with ν=5 is a pragmatic alternative. Multivariate extensions (joint distributions, covariance matrices, independence). Weak Law of Large Numbers, Central Limit Theorem, and unbiased estimators.

**Key Ch 2 insight**: Two toy models illustrate diversification. Model 1 (iid returns) suggests you can diversify away *all* risk. Model 2 (returns = β×market + idiosyncratic) shows you can only diversify away idiosyncratic risk — systematic risk remains. This motivates the passive investing view ("just own the market") and sets up the question: what justifies active management?

**My approach**: I'm comfortable with the calculus and linear algebra; I'm here for the finance intuition and the book's arc toward coherent risk measures. I skipped the Ch 2 exercises (PDF integration drills) since later chapters use distributions as tools rather than deriving them.

**Ch 3 covered**: Covariance as an inner product (giving Cauchy-Schwarz bounds on correlation). CAPM revisited: β = ρ(σ_r/σ_m), derived via inner product machinery. Key insight: correlation measures *linear* dependence only — a perfect quadratic relationship can show ρ ≈ 0 if positive and negative branches cancel.

**Copulas**: Sklar's theorem lets you separate any joint distribution into (1) a copula (pure dependence structure in [0,1]ᴺ space) and (2) arbitrary marginals. The probability integral transform (F(X) ~ U[0,1]) is the bridge. Gaussian copula infamously underestimated tail dependence in 2008 CDO crisis — Student-t copulas better capture "extremes cluster" behavior.

**Eigendecomposition of Σ**: Eigenportfolios (eᵢ'X) are uncorrelated and variance-ranked (Var = λᵢ). This is just PCA on returns. Empirically: one dominant eigenvalue (the market), a few sector factors, then a bulk of near-zero eigenvalues that are mostly estimation noise (random matrix theory). Ill-conditioning (large λ_max/λ_min) causes optimization to blow up on those noisy directions.

**Key Ch 3 insight**: The largest eigenportfolio is a proxy for "the market" — lending empirical support to simple CAPM. Correlations are unstable (regime-dependent) and spike in crises, which copula models must account for or fail catastrophically.


**Ch 4 covered**: OLS derived from first principles. Least-squares minimization of CAPM recovers the method-of-moments β̂, with the key assumption σ̂_{mε}=0 doing the heavy lifting. Adding an intercept α (Jensen's alpha) makes ε̄=0 and σ̂_{mε}=0 fall out automatically from first-order conditions — no explicit assumption needed. α≠0 implies statistical arbitrage; EMH says it shouldn't exist, but empirically it does (anomalies). A better model absorbs anomalies into new factors, but more factors → overfitting/factor zoo (same disease, individual vs collective strains).

**CAPM implications vs reality**: Theory says E(r) increases with β, σ_r, and ρ (risk is compensated). Empirical double-sorts on volatility/correlation show the opposite — low-vol stocks have *better* risk-adjusted returns (the low-volatility anomaly). CAPM fails as a predictive model but remains useful descriptive machinery.

**MLE → OLS**: Maximum likelihood with iid normal errors produces the same estimators as least squares, justifying the squared-error criterion.

**OLS in matrix form**: Y=Xβ+ε under Gauss-Markov assumptions. β̂=(X'X)⁻¹X'Y is unbiased. Hat matrix H projects Y onto col(X); residuals ε̂=(I-H)Y land in the orthogonal complement — a clean binary split (vs SVD's spectrum of singular values). Estimated residuals become correlated (Cov(ε̂)=σ²(I-H)) even though true errors aren't, because projection into a lower-dimensional subspace couples them. With normality assumption: standardized β̂ᵢ follows Student t_{N-p} (because σ is estimated, not known).

**Hypothesis testing & model selection**: t-tests for individual coefficients, F-tests for nested submodels (comparing RSS). Forward selection, backward elimination, stepwise regression. Corporate bond spread example illustrates forward selection with fundamental/technical factors — and contrasts with equities: in credit, risk *is* compensated.

**GLS**: When Cov(ε)=V≠σ²I, transform via V^{-1/2} to recover standard OLS structure. **Condition number** κ(X) bounds how perturbations in Ŷ amplify into perturbations in β̂ — collinearity means ill-conditioning, motivating variable selection.

**Key Ch 4 insight**: The OLS/projection framework is elegant and general, but the CAPM it was built to serve fails empirically. The machinery (α, β, F-tests, variable selection) survives as tools for model comparison and anomaly detection, even as the specific model assumptions don't hold.

**Ch 5 covered**: Math preliminaries for optimization. Gradient (direction of steepest ascent, via directional derivative + Cauchy-Schwarz), Hessian (matrix of second partials, indicates local curvature), Jacobian (row-stacked gradients of vector-valued functions; "Jacobian of the gradient is the Hessian"). Product rule for vector functions and gradient/Hessian of quadratic forms. Big O and little o notation for growth-rate classification. Multivariate Taylor expansions (first and second order) in several formulations — the intermediate-value form f(x+δ) = f(x) + ∇f(x+tδ)'δ for t∈(0,1) is used heavily in later proofs. Convex functions: three equivalent characterizations — (1) definition (function lies below chords), (2) for differentiable f, function lies above all tangent hyperplanes, (3) for C² f, Hessian is PSD everywhere. Proofs of (1)⟺(2) and (1)⟺(3) given. Dan skimmed without comments. 

**Ch 6 covered**: Unconstrained optimization. Global/local/strict minimizers defined. First-order necessary condition: ∇f*=0 (stationary point). Second-order necessary: ∇²f* ⪰ 0. Second-order sufficient: ∇f*=0 and ∇²f* ≻ 0 ⟹ local min. All proved via Taylor expansion + contradiction. OLS and GLS re-derived as unconstrained quadratic minimization (gradient=0 recovers familiar β̂, Hessian is PSD confirming optimality — but statistical properties of estimators are invisible in this purely optimization framing). Index tracking example: unconstrained least-squares replication of a market index produces good tracking (3 bps average error) but ~20% of weights are negative and positions as large as ±28% — motivating the need for constraints and eigenvalue-aware methods. Convex functions: any local min is global; stationarity alone is sufficient for global optimality. Newton's method: δᵏ = −(∇²fᵏ)⁻¹∇fᵏ, proved to converge quadratically (‖hᵏ⁺¹‖ = O(‖hᵏ‖²)) when started sufficiently close to x* with ∇²f* ≻ 0 and Lipschitz Hessian. Dan skimmed without comments. 

**Ch 7 covered**: Constrained optimization. Feasible sets, active sets, regular points (linearly independent active constraint gradients). Motivation via single equality constraint (gradient of f must be parallel to gradient of c at optimum) and single inequality constraint (interior: unconstrained; boundary: same as equality, plus complementarity). KKT conditions: ∇ₓℒ=0, feasibility, λᵢ≥0 for inequalities, complementarity λᵢcᵢ=0. Second-order necessary/sufficient conditions use the critical cone. Formal derivation of first-order conditions via inverse mapping theorem (change coordinates so constraints become zᵢ=0). Lagrange multipliers interpreted as sensitivity of f* to constraint perturbation. Convexity (f convex, cᵢ concave) makes KKT sufficient for global optimality.
