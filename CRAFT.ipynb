**Ch 3 covered**: Covariance as an inner product (giving Cauchy-Schwarz bounds on correlation). CAPM revisited: β = ρ(σ_r/σ_m), derived via inner product machinery. Key insight: correlation measures *linear* dependence only — a perfect quadratic relationship can show ρ ≈ 0 if positive and negative branches cancel.

**Copulas**: Sklar's theorem lets you separate any joint distribution into (1) a copula (pure dependence structure in [0,1]ᴺ space) and (2) arbitrary marginals. The probability integral transform (F(X) ~ U[0,1]) is the bridge. Gaussian copula infamously underestimated tail dependence in 2008 CDO crisis — Student-t copulas better capture "extremes cluster" behavior.

**Eigendecomposition of Σ**: Eigenportfolios (eᵢ'X) are uncorrelated and variance-ranked (Var = λᵢ). This is just PCA on returns. Empirically: one dominant eigenvalue (the market), a few sector factors, then a bulk of near-zero eigenvalues that are mostly estimation noise (random matrix theory). Ill-conditioning (large λ_max/λ_min) causes optimization to blow up on those noisy directions.

**Key Ch 3 insight**: The largest eigenportfolio is a proxy for "the market" — lending empirical support to simple CAPM. Correlations are unstable (regime-dependent) and spike in crises, which copula models must account for or fail catastrophically.


**Ch 4 covered**: OLS derived from first principles. Least-squares minimization of CAPM recovers the method-of-moments β̂, with the key assumption σ̂_{mε}=0 doing the heavy lifting. Adding an intercept α (Jensen's alpha) makes ε̄=0 and σ̂_{mε}=0 fall out automatically from first-order conditions — no explicit assumption needed. α≠0 implies statistical arbitrage; EMH says it shouldn't exist, but empirically it does (anomalies). A better model absorbs anomalies into new factors, but more factors → overfitting/factor zoo (same disease, individual vs collective strains).

**CAPM implications vs reality**: Theory says E(r) increases with β, σ_r, and ρ (risk is compensated). Empirical double-sorts on volatility/correlation show the opposite — low-vol stocks have *better* risk-adjusted returns (the low-volatility anomaly). CAPM fails as a predictive model but remains useful descriptive machinery.

**MLE → OLS**: Maximum likelihood with iid normal errors produces the same estimators as least squares, justifying the squared-error criterion.

**OLS in matrix form**: Y=Xβ+ε under Gauss-Markov assumptions. β̂=(X'X)⁻¹X'Y is unbiased. Hat matrix H projects Y onto col(X); residuals ε̂=(I-H)Y land in the orthogonal complement — a clean binary split (vs SVD's spectrum of singular values). Estimated residuals become correlated (Cov(ε̂)=σ²(I-H)) even though true errors aren't, because projection into a lower-dimensional subspace couples them. With normality assumption: standardized β̂ᵢ follows Student t_{N-p} (because σ is estimated, not known).

**Hypothesis testing & model selection**: t-tests for individual coefficients, F-tests for nested submodels (comparing RSS). Forward selection, backward elimination, stepwise regression. Corporate bond spread example illustrates forward selection with fundamental/technical factors — and contrasts with equities: in credit, risk *is* compensated.

**GLS**: When Cov(ε)=V≠σ²I, transform via V^{-1/2} to recover standard OLS structure. **Condition number** κ(X) bounds how perturbations in Ŷ amplify into perturbations in β̂ — collinearity means ill-conditioning, motivating variable selection.

**Key Ch 4 insight**: The OLS/projection framework is elegant and general, but the CAPM it was built to serve fails empirically. The machinery (α, β, F-tests, variable selection) survives as tools for model comparison and anomaly detection, even as the specific model assumptions don't hold.

**Ch 8 covered**: Mean-Variance Optimization (MVO). Merton's standard problem: minimize portfolio variance $\frac{1}{2}w'\Sigma w$ subject to a target expected return $\mu'w = \mu^*$ and full allocation $\mathbf{1}'w = 1$. This is a QPLC with a unique closed-form solution. The optimal weights $w^* = \lambda_1^* \Sigma^{-1}\mu + \lambda_2^* \Sigma^{-1}\mathbf{1}$ are determined by three scalars $a, b, c$ (quadratic forms involving $\Sigma^{-1}$, $\mu$, and $\mathbf{1}$). The optimal variance is quadratic in $\mu^*$, tracing a parabola — the **efficient frontier** — in volatility-return space. The vertex is the **global minimum variance optimal (GMVO)** portfolio at $\underline{\mu} = a/c$.

**Mutual fund separation theorem**: Because optimal weights are *affine* in $\mu^*$ ($w^* = \alpha + \gamma\mu^*$), any frontier portfolio can be replicated as a blend of any two other frontier portfolios. This is a massive dimensionality reduction: N assets → 2 "mutual funds."

**Capital Market Line (CML)**: Adding a risk-free asset linearizes the efficient frontier. The optimization (now with one constraint: $(\\mu - r_f)'w = \mu^* - r_f$) yields a *linear* relationship $\mu^* = r_f + \sigma\sqrt{cr_f^2 - 2ar_f + b}$. This line is tangent to the risky-asset frontier; the tangent point is the **tangency portfolio**. A second mutual fund separation theorem follows: every investor's optimal portfolio is a blend of the risk-free asset and the tangency portfolio. Under CAPM assumptions (and the mild condition $\mu_T = \mu_m$), the tangency portfolio *is* the market portfolio.

**Geometric derivation of CAPM**: By forming convex combinations $\Pi_\theta = \theta\Pi + (1-\theta)\Pi_T$ and evaluating $\partial\mathbb{E}(r_\theta)/\partial\sigma_\theta$ at $\theta=0$ via the chain rule, Chris shows that the CML slope forces the CAPM relation $\mu_0 - r_f = \beta_\Pi(\mu_T - r_f)$ — with no distributional assumption on the idiosyncratic error.

**Sharpe ratio optimization** (§8.2): Maximizing the Sharpe ratio $w'\mu - r_f)/\sqrt{w'\Sigma w}$ doesn't fit standard QP form. A change of variables ($y = \kappa w$, $\kappa = ((\\mu - r_f)'w)^{-1} > 0$) converts it to a QP with conic (then linear) constraints.

**Portfolio updates** (§8.3): Rewriting $w = w_0 + \Delta$ converts MVO into a QP in $\Delta$ with a linear term $w_0'\Sigma\Delta$. Turnover constraints $\sum|\Delta_i| \le \tau$ are handled by splitting $\Delta = \Delta_+ - \Delta_-$ (tripling dimension to $3N$). Slack variables address potential infeasibility.

**Equivalence of formulations** (§8.4): Min-variance-given-return (8.41) and max-return-given-variance (8.42) have identical solutions, proved by showing KKT multipliers map between the two problems via $\tilde{\delta}^* = 1/\delta^*$. A third form, $\min \frac{1}{2}w'\Sigma w - \gamma\mu'w$, parameterizes the frontier explicitly via risk appetite $\gamma$.

**Pitfalls** (§8.5): Unconstrained MVO is an "error maximizer." Near-zero eigenvalues of $\Sigma$ (estimation noise) cause the optimizer to take huge leveraged bets on low-eigenvalue eigenportfolios it perceives as riskless. The result: extreme long/short concentrations, the opposite of diversification. Remedies: constraints (which implicitly modify $\Sigma$), shrinkage estimators, factor models, random matrix theory — previewing chapters 9–10.

**Ch 9 covered**: Coherent measures of risk — four axioms that any "reasonable" risk measure should satisfy: (1) Positive Homogeneity (leverage scales risk linearly), (2) Translation Invariance (adding a risk-free position reduces risk by its certain profit), (3) Monotonicity (a portfolio that always loses less is less risky), and (4) Subadditivity (diversification can't increase risk beyond the sum of parts — the triangle inequality). Sign convention: positive $r$ = loss throughout.

**Volatility** satisfies homogeneity and subadditivity (via Cauchy-Schwarz), but fails translation invariance and monotonicity because it's location-independent. The fix $\rho_\gamma(\Pi) = \mathbb{E}(\Pi'r) + \gamma\|\Pi\|_\Sigma$ adds location dependence and is coherent for elliptical distributions, where $\gamma$ maps directly to a percentile of the loss distribution via $\Phi(\gamma)$.

**VaR** ($\beta$-quantile of the loss distribution) is intuitive and regulatory-standard, but fails subadditivity in general — the bond default example shows two individually safe positions (95-VaR = −$1 each) that combine into a riskier portfolio (95-VaR = $8.24). VaR's fatal flaw: it marks the *threshold* of the danger zone but is blind to what happens inside it.

**CVaR** (expected loss conditional on being in the worst $1-\beta$ tail) fixes this. It's always ≥ VaR and is a coherent risk measure. Three equivalent formulations: (1) conditional expectation (9.6), (2) average of all $\tilde\beta$-VaR for $\tilde\beta \in [\beta, 1]$ (9.7), and (3) the Rockafellar-Uryasev minimization $\min_\alpha G_\beta(\Pi, \alpha)$ (9.9), where $G_\beta$ is convex in $\alpha$ and jointly convex in $(\Pi, \alpha)$ — the key property enabling LP implementation.

**Drawdown measures** add a time/path dimension. Maximum drawdown and average drawdown capture serial clustering of losses but fail translation invariance and subadditivity. $\beta$-CDaR (conditional drawdown-at-risk) is the CVaR analogue applied to the drawdown distribution — coherent except for translation invariance, fixable by adding an expectation term (exercise 6 pattern).

**Implementation as LP constraints** (§9.2): All risk measures are made operational via auxiliary variables that linearize the $[\cdot]_+$ operator. CVaR constraint (9.23): introduce slack $z_k \ge \max(0, \Pi'r_k - \alpha)$, bound $\alpha + \frac{1}{1-\beta}\frac{1}{K}\sum z_k \le \omega$. As a byproduct, $\alpha^*$ yields the $\beta$-VaR. Drawdown constraints (9.25): auxiliary $z_k$ tracks running peak via $z_k \ge z_{k-1}$, $z_k \ge -\Pi'R_k$. CDaR constraints (9.27): layer CVaR-style slacks $u_k$ on top of the drawdown tracking variables. All formulations are convex, fitting into the LP/QP machinery of Ch 7.

**Sampling** (§9.2.3): Copulas (Ch 3) generate the required loss scenarios. Student $t_5$ copula preferred for equities (fat tails). For equity + options portfolios, the Taylor-expansion pricing (delta-gamma) makes portfolio losses non-elliptical even when equity returns are — precisely where CVaR adds value over variance as a risk measure, since variance-based methods assume elliptical portfolio loss distributions.

**Key Ch 9 insight**: The chapter builds a hierarchy: volatility (not coherent) → VaR (coherent only for elliptical, fails subadditivity in general) → CVaR (coherent, penalizes tails, LP-implementable) → CDaR (adds path-dependence). The Rockafellar-Uryasev $G_\beta$ formulation is the linchpin — its joint convexity in $(\Pi, \alpha)$ is what makes CVaR constraints practical. This directly answers the Ch 8 question: how to optimize portfolios with risk measures that actually respect fat tails and diversification, rather than relying on variance (which assumes away exactly the dangers that matter).

**Ch 10 covered**: Covariance modifications — three strategies for replacing the noisy sample covariance $\hat{\Sigma}$ with a better-behaved (biased but stabler) estimator, all trading bias for reduced estimation error.

**Factor models** (§10.1): Impose full structure via $\Sigma = B\Omega_f B' + D$. Time-series approach (observed factors, estimate loadings via OLS), cross-sectional approach (observed loadings, estimate factors via GLS), and statistical approach (PCA eigenportfolios — literally truncated SVD of $\hat{\Sigma}$). All reduce the $\frac{N(N+1)}{2}$ estimation problem to $O(NK)$ parameters when $K \ll N$.

**Shrinkage** (§10.2–10.4): Convex combination $\Sigma_s = (1-\alpha)\hat{\Sigma} + \alpha\hat{\Omega}$ of sample and structured target. Ledoit-Wolf derive the optimal $\alpha^*$ minimizing expected Frobenius-norm loss $\mathbb{E}\|\Sigma_s - \Sigma\|_F^2$. Two worked targets: constant correlation ($\hat{\Omega}$ uses average pairwise $\bar{\rho}$) and CAPM single-factor. The consistent estimator $\hat{\alpha}^*$ decomposes into terms $p$, $q$, $c$ computable from data — $q$ depends on the target choice.

**Constraint-implied modification** (§10.5): The chapter's distinctive contribution. Solving a constrained MVO implicitly modifies $\hat{\Sigma}$ into $\tilde{\Sigma} = \hat{\Sigma} + \Delta$, where $\Delta$ is built from the KKT multipliers and return means. Three propositions: (1) $\omega^*$ of the constrained problem also solves the *unconstrained* mean-variance problem with $\tilde{\Sigma}$; (2) when $\tilde{\Sigma}$ is invertible, it solves a constrained maximum likelihood problem; (3) $\tilde{\Sigma} \succeq 0$ whenever $S \succeq 0$. This explains why "imposing the wrong constraints helps" — constraints act as implicit shrinkage. The value-constraint example (§10.5.1) shows deliberate tilting: requiring $\geq\nu\%$ in high-EBIT/EV stocks produces a $\tilde{\Sigma}$ whose minimum-variance portfolio has statistically significant value factor exposure (Table 10.1).

**Key Ch 10 insight**: All three approaches are forms of regularization against the "error maximizer" problem from Ch 8. Factor models ≈ truncated SVD; shrinkage ≈ ridge regression; constraints implicitly shrink by zeroing out directions the optimizer would exploit. Equal weighting is the extreme case ($\alpha \to 1$): trust zero signal from $\hat{\Sigma}$, which empirically often wins because it completely sidesteps estimation error. For retail investors, the practical takeaway is that simple structural choices (equal weighting, sector constraints) provide much of the same stabilization that professionals achieve with Ledoit-Wolf and custom factor tilts.
